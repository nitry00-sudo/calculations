```python
import numpy as np
```

```python
data = np.loadtxt('/Users/charlie/Downloads/MnPc-O-Cu Band Structure.csv',delimiter=',',usecols=[0,1])
```

```python
clean_data = data[~np.any(data > 1e+300, axis=1)]
x_coord = np.unique(clean_data[:, 0])
y_coord = clean_data[:, 1:]
print(x_coord)
```

```python
rows = np.shape(x_coord)[0]
columns = np.shape(y_coord)[0]//rows
print(rows, columns)
```

```python
Y = y_coord.reshape(rows, columns)
```

```python
final_data = np.insert(Y,0,x_coord,axis=1)
print(final_data)
```
